## Root-Finding Method: Newton-Raphson
### $$ x_{new} = x_{old} - \frac{f(x_{old})}{f'(x_{old})} $$

```python
import numpy as np
import pandas as pd
```

```python
def newton(x0, tol):
    x_old = x0
    i = 1

    if dfunc(x0) == 0:
        print("This method doesn't work! The derivative is zero at the initial guess.")
        print("please change the intiat guess x0")
        return None
        
    x_new= x_old - func(x_old) / dfunc(x_old)
    Err = (abs(x_new - x_old)/ abs(x_new)) * 100
    print('{:2d} {:8.4f} {:8.4f} {:8.4f} {:8.4f} {:8.4f}' .format(
        i, x_old, func(x_old), dfunc(x_old), x_new, Err))

    while Err>tol:
        x_old = x_new
        x_new = x_old - func(x_old)/ dfunc(x_old)
        Err =(abs(x_new - x_old) / abs(x_new)) * 100
        i = i+1

        print('{:2d} {:8.4f} {:8.4f} {:8.4f} {:8.4f} {:8.4f}' .format(
        i, x_old, func(x_old), dfunc(x_old), x_new, Err))

        if func(x_new) == 0:
            return x_new

    return x_new
    
```

```python
def newton(x0, tol):
    rows = []
    i = 1

    while True:
        if dfunc(x0) == 0:
            print("Turunan bernilai nol, ganti tebakan awal x0.")
            return None

        x1 = x0 - func(x0) / dfunc(x0)

        err = abs(x1 - x0) / abs(x1) * 100 if x1 != 0 else np.inf

        rows.append([i, x0, func(x0), dfunc(x0), x1, err])

        if err <= tol or func(x1) == 0:
            break

        x0 = x1
        i += 1

    df = pd.DataFrame(
        rows,
        columns=[
            "Iterasi",
            "x_old",
            "f(x_old)",
            "f'(x_old)",
            "x_new",
            "Error (%)"
        ]
    )

    display(
        df.style.format({
            "x_old": "{:.5f}",
            "f(x_old)": "{:.5f}",
            "f'(x_old)": "{:.5f}",
            "x_new": "{:.5f}",
            "Error (%)": "{:.5f}"
        })
        .hide(axis="index")
        .set_caption("Tabel Hasil Pencarian Akar dengan Metode Newton-Raphson")
        .set_properties(**{
            "text-align": "center",
            "padding": "8px 12px"
        })
        .set_table_styles([
            {
                "selector": "caption",
                "props": [
                    ("font-size", "14px"),
                    ("font-weight", "bold"),
                    ("padding", "10px")
                ]
            },
            {
                "selector": "th",
                "props": [
                    ("background-color", "#215997"),
                    ("color", "white"),
                    ("text-align", "center"),
                    ("padding", "10px")
                ]
            },
            {
                "selector": "td",
                "props": [
                    ("padding", "8px 12px")
                ]
            },
            {
                "selector": "tbody tr:nth-child(even)",
                "props": [
                    ("background-color", "#F3F5FA")
                ]
            }
        ])
    )

    return x1
```

### Soal 1
### Fungsi: $$f(x) = x\sin(x) - e^x$$
#### Turunan pertama: $$f'(x) = \sin(x) + x\cos(x) - e^x$$
## x0=0.5, tol= 0.00001

```python
def func(x):
    return x * np.sin(x) - np.exp(x)
def dfunc(x):
    return np.sin(x) + x*np.cos(x) - np.exp(x)
```

```python
print("Iter  x_old  f(x_old)  f'(x_old)  x_new  Error(%)")
print("================================================")

root = newton(0.5, 0.00001)

print("=================================================")
print("The root of f(x) is", round(root, 6))
```

### Soal 2
### Fungsi: $$f(x) = 1 - x + \sin(x)$$
#### Turunan pertama: $$f'(x) = -1 + \cos(x)$$
## x0=0.5, tol= 0.00001

```python
def func(x):
    return 1 - x + np.sin(x)
def dfunc(x):
    return -1 + np.cos(x)
```

```python
print("Iter  x_old  f(x_old)  f'(x_old)  x_new  Error(%)")
print("================================================")

root = newton(0.5, 0.00001)

print("=================================================")
print("The root of f(x) is", round(root, 6))
```

### Soal 3

#### Turunan pertama: $$e^x$$
## x0=2, tol= 0.0001

```python
def func(x):
    return np.exp(x) - 1
def dfunc(x):
    return np.exp(x)
```

```python
print("Iter  x_old  f(x_old)  f'(x_old)  x_new  Error(%)")
print("================================================")

root = newton(2, 0.0001)

print("=================================================")
print("The root of f(x) is", round(root, 6))
```
