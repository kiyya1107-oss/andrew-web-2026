```python
def bisection(a, b, t, max_iteration):
    print("iter     a      b       c      f(a)      f(b)     f(c)      Err")
    print("------------------------------------------------------------------------------------------")
    if func(a)*func(b)>0:
        print("Error!")
        print("Change a and b")
    else:
        for i in range(1, max_iteration):
            c= (a+b)/2
            Err= abs(a-b)
            print("{:2d} {:8.4f} {:8.4f} {:8.4f} {:8.4f} {:8.4f} {:8.4f} {:8.4f}"
                .format(i, a, b, c, func(a), func(b), func(c), Err))
            if func(c)==0:
                return c
            elif Err<t:
                break

            elif func(a)*func(c)<0:
                b=c
            else:
                a=c
        return c
    print("------------------------------------------------------------------------------------------")
    print(f"The root of f(x) is", round(c,5))
```

## Function 1
### $$ f(x) = x * math.sin(x) - math.exp(x), [-2,3], t= 0.001 $$

```python
import math

def func(x):
    return x * math.sin(x) - math.exp(x)

c = bisection (-2,3,0.001,50)
print("------------------------------------------------------------------------------------------")
print(f"The root of f(x) is", round(c,5))
```
